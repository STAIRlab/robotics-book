### Estimating Expected Cost of Robot Paths

```python
np.random.seed(0)
# Path cost distributions (mean, std)
paths = [(10, 2), (12, 1), (9, 3)]
N = 1000

for i, (mu, sigma) in enumerate(paths):
    samples = np.random.normal(mu, sigma, N)
    expected_cost = np.mean(samples)
    print(f"Path {i+1}: Expected Cost = {expected_cost:.2f}")
```

### Estimating pi by Monte Carlo integration

Sample points in the unit square and count those inside a quarter circle.

```python
np.random.seed(42)
def estimate_pi(num_samples=1000000):
    x = np.random.rand(num_samples)
    y = np.random.rand(num_samples)
    inside_circle = (x**2 + y**2) <= 1
    pi_estimate = 4 * np.sum(inside_circle) / num_samples
    return pi_estimate

# Run the simulation
pi_value = estimate_pi()
print(f"Estimated pi: {pi_value:.6f}")
```
