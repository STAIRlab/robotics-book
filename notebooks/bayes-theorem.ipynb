## Interpretation: How information changes the belief

Before the LiDAR measurement, the robot assigns equal prior probabilities to three candidate locations.

The measurement event has likelihoods of 0.8, 0.4, and 0.1 under locations L1, L2, and L3, respectively.

Bayesian updating changes the probabilities as follows:

- L1: 0.333 to 0.615.
- L2: 0.333 to 0.308.
- L3: 0.333 to 0.077.

The likelihoods are illustrative event probabilities, not measurements from a field-deployed rover. The posterior probabilities are conditional on the three candidate locations being mutually exclusive and exhaustive.
